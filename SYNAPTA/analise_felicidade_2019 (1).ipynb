## 5. Análise de correlação

Nesta etapa, foi investigada a relação entre a pontuação de felicidade e diferentes indicadores econômicos e sociais.

Foram analisadas as seguintes relações:

- PIB per capita e pontuação de felicidade: correlação de 0,7939.
- Apoio social e pontuação de felicidade: correlação de 0,7771.
- Expectativa de vida saudável e pontuação de felicidade: correlação de 0,7799.

Os três indicadores apresentaram correlações positivas fortes com a pontuação de felicidade.

A generosidade apresentou uma correlação linear muito fraca, de 0,0758.

Também foi construída uma matriz de correlação para observar as relações entre todas as variáveis numéricas.

É importante destacar que correlação não significa causalidade. Os resultados indicam associações entre os indicadores, mas não comprovam que um fator seja responsável diretamente pelo aumento da felicidade.



onde fica isso?

```python
df["Score"].corr(df["Social support"])
```

```python
df[["Country or region", "Score", "Healthy life expectancy"]].head(10)
```

```python
df["Score"].corr(df["Healthy life expectancy"])
```

```python
correlacao = df.corr(numeric_only=True)

correlacao
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Matriz de correlação - World Happiness Report 2019")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Social support"], df["Score"])

plt.xlabel("Apoio social")
plt.ylabel("Pontuação de felicidade")
plt.title("Apoio social x Felicidade - 2019")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Healthy life expectancy"], df["Score"])

plt.xlabel("Expectativa de vida saudável")
plt.ylabel("Pontuação de felicidade")
plt.title("Expectativa de vida saudável x Felicidade - 2019")

plt.show()
```
